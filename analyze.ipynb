```python
import pandas as pd

# Show all columns (no "..." in the middle)
pd.set_option("display.max_columns", None)
pd.set_option("display.width", None)
pd.set_option("display.max_colwidth", 50)

# Load results from batch run
results_path = "/home/lando/Documents/GitHub/found-integration-testing/results/comprehensive/baseline.csv"
df = pd.read_csv(results_path)

print(f"Loaded {len(df)} rows. Columns: {list(df.columns)}")
```

```python
from limb.simulation.analysis.metrics import fill_pixel_metrics

# Ensure metric columns exist so fill_pixel_metrics can fill them
metric_cols = [
    "true_x_centroid", "true_y_centroid", "true_r_apparent",
    "out_x_centroid", "out_y_centroid", "out_r_apparent",
]
for c in metric_cols:
    if c not in df.columns:
        df[c] = float("nan")

df = fill_pixel_metrics(df)

# Show position, orientation, output position, and filled centroid/r
cols = [
    "true_pos_x", "true_pos_y", "true_pos_z",
    "qx", "qy", "qz", "qw",
    "out_pos_x", "out_pos_y", "out_pos_z",
    "true_x_centroid", "true_y_centroid", "true_r_apparent",
    "out_x_centroid", "out_y_centroid", "out_r_apparent",
]
cols = [c for c in cols if c in df.columns]
df[cols]
```

```python
import numpy as np

# Pixel error = Euclidean distance between true and output centroid (pixels)
valid = (
    df["out_x_centroid"].notna()
    & df["out_y_centroid"].notna()
    & df["true_x_centroid"].notna()
    & df["true_y_centroid"].notna()
)
pixel_error = np.sqrt(
    (df.loc[valid, "out_x_centroid"] - df.loc[valid, "true_x_centroid"]) ** 2
    + (df.loc[valid, "out_y_centroid"] - df.loc[valid, "true_y_centroid"]) ** 2
)
df_with_error = df.loc[valid].copy()
df_with_error = df_with_error.assign(pixel_error=pixel_error.values)

top10_pixel_errors = df_with_error.nlargest(10, "pixel_error")

# Position error = Euclidean distance between true and output position (meters)
valid_pos = (
    df["true_pos_x"].notna() & df["true_pos_y"].notna() & df["true_pos_z"].notna()
    & df["out_pos_x"].notna() & df["out_pos_y"].notna() & df["out_pos_z"].notna()
)
position_error = np.sqrt(
    (df.loc[valid_pos, "out_pos_x"] - df.loc[valid_pos, "true_pos_x"]) ** 2
    + (df.loc[valid_pos, "out_pos_y"] - df.loc[valid_pos, "true_pos_y"]) ** 2
    + (df.loc[valid_pos, "out_pos_z"] - df.loc[valid_pos, "true_pos_z"]) ** 2
)
df_with_pos_error = df.loc[valid_pos].copy()
df_with_pos_error = df_with_pos_error.assign(position_error=position_error.values)

top10_position_errors = df_with_pos_error.nlargest(10, "position_error")

# Mean difference in norm (position: ‖out - true‖ in m; pixel: ‖out_centroid - true_centroid‖ in px)
mean_position_error_m = df_with_pos_error["position_error"].mean()
mean_pixel_error_px = df_with_error["pixel_error"].mean()
print(f"Mean position error (norm of difference, m): {mean_position_error_m:.4f}")
print(f"Mean pixel error (norm of difference, px):    {mean_pixel_error_px:.4f}")

cols = [
    "true_pos_x", "true_pos_y", "true_pos_z",
    "qx", "qy", "qz", "qw",
    "out_pos_x", "out_pos_y", "out_pos_z",
    "true_x_centroid", "true_y_centroid", "true_r_apparent",
    "out_x_centroid", "out_y_centroid", "out_r_apparent",
]
cols = [c for c in cols if c in df.columns]
print("Top 10 pixel errors (centroid residual, px):")
display(top10_pixel_errors[cols + ["pixel_error"]])
print("\nTop 10 position errors (3D residual, m):")
display(top10_position_errors[cols + ["position_error"]])
```

```python
import numpy as np
from limb.simulation.analysis.plot import (
    plot_centroid_residuals_in_illumination_frame,
    plot_radius_residuals_vs_range,
)

# True range (m) from true position
ranges_m = np.sqrt(
    df["true_pos_x"] ** 2 + df["true_pos_y"] ** 2 + df["true_pos_z"] ** 2
).values

# Residuals: estimated minus true (only use rows with valid out_*)
valid = (
    df["out_x_centroid"].notna()
    & df["out_y_centroid"].notna()
    & df["out_r_apparent"].notna()
    & df["true_x_centroid"].notna()
    & df["true_y_centroid"].notna()
    & df["true_r_apparent"].notna()
)

# Cutoff: keep only range <= 20e6 m
RANGE_CUTOFF_M = 20e6
valid = valid & (ranges_m <= RANGE_CUTOFF_M)

ranges_valid = ranges_m[valid]
centroid_x_residual_px = (
    df.loc[valid, "out_x_centroid"] - df.loc[valid, "true_x_centroid"]
).values
centroid_y_residual_px = (
    df.loc[valid, "out_y_centroid"] - df.loc[valid, "true_y_centroid"]
).values
radius_residual_px = (
    df.loc[valid, "out_r_apparent"] - df.loc[valid, "true_r_apparent"]
).values

# Cut out outliers: keep points within 3 IQR of median for each residual
def mask_outliers_iqr(x, k=1.5):
    q1, q3 = np.nanpercentile(x, [25, 75])
    iqr = q3 - q1
    if iqr <= 0:
        return np.ones_like(x, dtype=bool)
    return (x >= q1 - k * iqr) & (x <= q3 + k * iqr)

inlier_x = mask_outliers_iqr(centroid_x_residual_px)
inlier_y = mask_outliers_iqr(centroid_y_residual_px)
inlier_r = mask_outliers_iqr(radius_residual_px)
inlier = inlier_x & inlier_y & inlier_r
ranges_valid = ranges_valid[inlier]
# Series key: FOV (deg) only — FOV_deg ≈ 2 * arctan((res * pixel_pitch) / (2 * focal_length)) * 180/pi
_res = df.loc[valid, "cam_x_resolution"].values
_fl = df.loc[valid, "cam_focal_length"].values
_pp = df.loc[valid, "cam_x_pixel_pitch"].values
fov_deg = np.degrees(2 * np.arctan((_res * _pp) / (2 * _fl)))
fov_deg = np.round(fov_deg, 1)
_pass_ids_raw = np.asarray(fov_deg, dtype=str) + "deg"
pass_ids = _pass_ids_raw[inlier].tolist()

# Camera type = resolution (e.g. "512x512", "1024x1024")
cam_type_res = (
    df.loc[valid, "cam_x_resolution"].astype(int).astype(str).values
    + "x"
    + df.loc[valid, "cam_y_resolution"].astype(int).astype(str).values
)
cam_types = np.unique(cam_type_res[inlier])

if valid.sum() < 3:
    print("Need at least 3 rows with valid centroid/r to plot; skipping.")
else:
    plot_df_full = df.loc[valid].iloc[inlier].copy()
    plot_df_full["pass_id"] = pass_ids

    for cam_type in cam_types:
        mask_cam = (cam_type_res[inlier] == cam_type)
        if mask_cam.sum() < 3:
            continue
        plot_df = plot_df_full.iloc[mask_cam].copy()

        centroid_1sigma_px = float(
            np.nanstd(
                np.r_[
                    centroid_x_residual_px[inlier][mask_cam],
                    centroid_y_residual_px[inlier][mask_cam],
                ]
            )
            or 1.0
        )
        radius_1sigma_px = float(
            np.nanstd(radius_residual_px[inlier][mask_cam]) or 1.0
        )

        # Plot centroid residuals (X/Y) vs range — one figure per camera type
        plot_centroid_residuals_in_illumination_frame(
            plot_df,
            centroid_1sigma_px=centroid_1sigma_px,
            target_label="Earth",
            save_path=None,
        )

        # Plot radius residual vs range — one figure per camera type
        plot_radius_residuals_vs_range(
            plot_df,
            radius_1sigma_px=radius_1sigma_px,
            target_label="Earth",
            save_path=None,
        )
```

```python
import numpy as np

# Norm of input (true) and output (estimated) positions per row
true_norm = np.sqrt(
    df["true_pos_x"] ** 2 + df["true_pos_y"] ** 2 + df["true_pos_z"] ** 2
).values
out_pos_valid = (
    df["out_pos_x"].notna() & df["out_pos_y"].notna() & df["out_pos_z"].notna()
)
out_norm = np.sqrt(
    df["out_pos_x"] ** 2 + df["out_pos_y"] ** 2 + df["out_pos_z"] ** 2
).values
out_norm = out_norm[out_pos_valid]

print("True position norm (m):")
print(f"  mean = {np.mean(true_norm):.2f}")
print(f"  std  = {np.std(true_norm):.2f}")
print("Output position norm (m), valid rows only:")
print(f"  mean = {np.mean(out_norm):.2f}")
print(f"  std  = {np.std(out_norm):.2f}")
```
